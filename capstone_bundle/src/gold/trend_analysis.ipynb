## **daily revenue**

```python
%run "/Workspace/capstone_project/capstone_project_cyntexa/capstone_bundle/src/gold/sales_summary"
```

```python
%sql
WITH daily_sales AS (

SELECT

sale_date,

round(SUM(total_amount), 2) AS daily_revenue

FROM identifier(:catalog).silver.sales_clean

GROUP BY sale_date

)

SELECT *

FROM daily_sales

ORDER BY sale_date;
```

## **7-Day Moving Average**

```python
%sql
create or replace view identifier(:catalog).gold.Seven_Day_moving_sales AS
WITH daily_sales AS (

SELECT

sale_date,

round(SUM(total_amount), 2) AS daily_revenue

FROM identifier(:catalog).silver.sales_clean

GROUP BY sale_date

)

SELECT

sale_date,

daily_revenue,

ROUND(

AVG(daily_revenue)

OVER(

ORDER BY sale_date

ROWS BETWEEN 6 PRECEDING AND CURRENT ROW

),

2

) AS moving_avg_7_day

FROM daily_sales

ORDER BY sale_date;
```

## **30-Day Moving Average**

```python
%sql
create or replace view identifier(:catalog).gold.sales_moving_avg_30_day AS
WITH daily_sales AS (

SELECT

sale_date,

SUM(total_amount) AS daily_revenue

FROM identifier(:catalog).silver.sales_clean

GROUP BY sale_date

)

SELECT

sale_date,

daily_revenue,

ROUND(

AVG(daily_revenue)

OVER(

ORDER BY sale_date

ROWS BETWEEN 29 PRECEDING AND CURRENT ROW

),

2

) AS moving_avg_30_day

FROM daily_sales

ORDER BY sale_date;
```

## **Running Total**

```python
%sql
create or replace view identifier(:catalog).gold.sales_running_total AS
WITH daily_sales AS (

SELECT

sale_date,

round(SUM(total_amount), 2) AS daily_revenue

FROM identifier(:catalog).silver.sales_clean

GROUP BY sale_date

)

SELECT

sale_date,

daily_revenue,

round(SUM(daily_revenue)

OVER(

ORDER BY sale_date

ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW

), 2)

AS running_total

FROM daily_sales

ORDER BY sale_date;
```

## **Cumulative Revenue %**

```python
%sql
create or replace view identifier(:catalog).gold.sales_cumulative_revenue_percentage AS
WITH daily_sales AS (

SELECT

sale_date,

round(SUM(total_amount), 2) AS daily_revenue

FROM identifier(:catalog).silver.sales_clean

GROUP BY sale_date

)

SELECT

sale_date,

daily_revenue,

round(SUM(daily_revenue)

OVER(

ORDER BY sale_date

ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW

), 2)

AS cumulative_revenue,

ROUND(

SUM(daily_revenue)

OVER(

ORDER BY sale_date

ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW

)

/

SUM(daily_revenue)

OVER()

*100,

2

)

AS cumulative_percentage

FROM daily_sales

ORDER BY sale_date;
```
