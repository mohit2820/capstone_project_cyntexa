```python
from customer_cleaning import clean_customers
from product_cleaning import clean_products
from sales_cleaning import clean_sales
from supplier_cleaning import clean_suppliers
```

```python
dbutils.widgets.text("catalog","dev")
catalog = dbutils.widgets.get("catalog")
```

```python
# Read Bronze table
customers_df = spark.table(f"{catalog}.bronze.customers_raw")

# Clean customers
customers_clean = clean_customers(customers_df)

# Save to Silver
customers_clean.write \
    .mode("overwrite") \
    .saveAsTable(f"{catalog}.silver.customers_clean")
```

```python
# Read Bronze
sales_df = spark.table(f"{catalog}.bronze.sales_raw")

# Clean Sales
sales_clean = clean_sales(sales_df)

# Write to Silver
sales_clean.write \
    .mode("overwrite") \
    .saveAsTable(f"{catalog}.silver.sales_clean")
```

```python
# Read Bronze
suppliers_df = spark.table(f"{catalog}.bronze.suppliers_raw")

# Clean Suppliers
suppliers_clean = clean_suppliers(suppliers_df)

# Write to Silver
suppliers_clean.write \
    .mode("overwrite") \
    .saveAsTable(f"{catalog}.silver.suppliers_clean")
```

```python
# Read Bronze
products_df = spark.table(f"{catalog}.bronze.products_raw")

# Sales Df
sales_df = spark.table(f"{catalog}.silver.sales_clean")

# Clean Products
products_clean = clean_products(products_df,sales_df)

# Write to Silver
products_clean.write \
    .mode("overwrite") \
    .saveAsTable(f"{catalog}.silver.products_clean")
```
