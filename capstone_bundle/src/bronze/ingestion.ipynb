```python
%sql
CREATE CATALOG IF NOT EXISTS dev;
CREATE CATALOG IF NOT EXISTS uat;
CREATE CATALOG IF NOT EXISTS prod;
```

```python
dbutils.widgets.text("catalog" , "dev")
catalog = dbutils.widgets.get("catalog")
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).bronze;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).silver;
CREATE SCHEMA IF NOT EXISTS IDENTIFIER(:catalog).gold;
```

```python
%sql
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.customers_raw;
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.products_raw;
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.sales_raw;
CREATE TABLE IF NOT EXISTS IDENTIFIER(:catalog).bronze.suppliers_raw;
```

```python
%sql
create volume if not exists IDENTIFIER(:catalog).bronze.raw;
    
```

```python
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw/customers")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw/products")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw/sales")
dbutils.fs.mkdirs(f"/Volumes/{catalog}/bronze/raw/suppliers")
```

```python
%sql

COPY INTO IDENTIFIER(:catalog).bronze.sales_raw
FROM '/Volumes/' :catalog '/bronze/raw/sales/'
FILEFORMAT = CSV
FORMAT_OPTIONS ('header' = 'true')
COPY_OPTIONS('mergeSchema' = 'true')

```
