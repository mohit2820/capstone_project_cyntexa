```python
dbutils.widgets.text("catalog", "dev")
catalog = dbutils.widgets.get("catalog")
```

```python
customers = spark.table(f"{catalog}.silver.customers_clean")
customers.createOrReplaceTempView("customer_source")
```

```python
%sql
MERGE INTO IDENTIFIER(:catalog).silver.customers_clean AS target

USING customer_source AS source

ON target.customer_id = source.customer_id

WHEN MATCHED THEN
  UPDATE SET *

WHEN NOT MATCHED THEN
  INSERT (
    customer_id,
    name,
    email,
    city,
    state,
    signup_date
  )
  VALUES (
    source.customer_id,
    source.name,
    source.email,
    source.city,
    source.state,
    source.signup_date
);
```
